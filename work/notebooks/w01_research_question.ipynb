## 2. The question: decision, action, cost of a wrong call

**Who's the decision?** An editor deciding which pages to put work into this week.

**What do they act on?** A ranked list — "these 20 pages look like they're heading down, check them first."

**Cost of a wrong call?** Two ways to be wrong: you waste time fixing a page that was fine (annoying but cheap), or you miss a page that's quietly tanking and costs weeks of traffic (expensive).

**Why ML over a rule?** Decline isn't just "impressions dropped" — it's position slipping, CTR changing, engagement thinning. The shape is different for every page. A model can see the pattern across many pages in a way a single threshold can't.

```python
# One-paragraph frame:
# For editors and content strategists, deciding which pages to prioritize for refreshes,
# we will build a classification model from historical search and engagement data,
# predicting whether a page will decline (trend_direction == "down"),
# measured by precision@K (for the review queue) and ROC-AUC.
# A wrong call costs wasted editor hours (FP) or a missed decline that compounds (FN).
# A plain rule isn't enough because decline patterns are multi-signal —
# position drift + volume changes + engagement erosion — no single threshold catches them all.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import os

repo_root = os.path.abspath(os.path.join(os.getcwd(), "..", ".."))
df = pd.read_csv(os.path.join(repo_root, "data", "raw", "content_refresh_anonymized.csv"))

print(f"Shape: {df.shape[0]:,} rows, {df.shape[1]} columns, {df['client_id'].nunique()} clients")
print()

# 1. Trend direction breakdown — 54% declining
trend_counts = df['trend_direction'].value_counts()
print("Trend direction breakdown:")
for label, count in trend_counts.items():
    print(f"  {label}: {count:>6,} ({count/len(df)*100:.1f}%)")
print()

# 2. Declining pages lost ~60% of impressions
declining = df[df['trend_direction'] == 'down']
other = df[df['trend_direction'] != 'down']

print("Median impressions — declining vs rest:")
print(f"  Declining:  last_30d = {declining['impressions_last_30d'].median():,.0f},  prev_30d = {declining['impressions_prev_30d'].median():,.0f}")
print(f"  Rest:       last_30d = {other['impressions_last_30d'].median():,.0f},  prev_30d = {other['impressions_prev_30d'].median():,.0f}")
print()

# 3. Scale: most clients have 500-2000 pages
client_sizes = df.groupby('client_id').size()
print(f"Pages per client — min: {client_sizes.min()}, median: {client_sizes.median():.0f}, max: {client_sizes.max()}")
```

## 4. Careful words: what I can and can't claim

What this *can* say: "Pages with these patterns tended to decline in our data" — directional, observed, useful for triage.

What it *can't* say: "This page will decline" as fact, or that anything caused it, or that any of this applies outside these 32 clients and this time window. No predicting Google, no claiming causality. Just a better-informed queue for a human editor.

```python
# Can claim:
#   - observed correlations between features and decline in this dataset
#   - directional scoring ("this page's pattern matches declining pages we've seen")
#   - decision-support rankings for editorial review queues
#
# Cannot claim:
#   - causal proof ("X causes decline")
#   - generalization beyond these 32 clients and this 90-day window
#   - predicting Google algorithm changes or AI citation behavior
#   - absolute predictions ("this page will decline") — only directional signals
```
